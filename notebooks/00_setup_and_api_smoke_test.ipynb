# 00 - Setup and API Smoke Test

This notebook verifies that the environment works and that we can reach the OpenFEC API.
We will load configuration, print the API key status, and fetch a small sample of 2026 Senate candidates.

```python
from pathlib import Path
import sys
from pprint import pprint

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.append(str(ROOT / "src"))

from midterm_money.config import settings
from midterm_money.fec_client import FECClient
import pandas as pd
```

```python
print("Project root:", ROOT)
print("OpenFEC base URL:", settings.FEC_BASE_URL)
print("Election cycle:", settings.ELECTION_CYCLE)
print("Office filter:", settings.OFFICE)
print("Using demo key:", settings.USE_DEMO_KEY)
if settings.USE_DEMO_KEY:
    print("Warning: Using DEMO_KEY. Replace with a real FEC_API_KEY in .env for production use.")
```

```python
client = FECClient()
sample_params = {
    "cycle": settings.ELECTION_CYCLE,
    "office": settings.OFFICE,
    "per_page": 5,
    "sort": "name",
}
response = client.get("/candidates/search/", params=sample_params)
print("Response keys:", list(response.keys()))
sample_results = response.get("results", [])
print("Number of sample rows:", len(sample_results))
df = pd.DataFrame(sample_results)
print("Columns returned by sample request:", df.columns.tolist())
df.head()
```
